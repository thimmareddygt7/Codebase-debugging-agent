# 02: AST Chunking Experiments

This notebook demonstrates parsing code files using Python's Abstract Syntax Tree (AST) to produce semantic function and class chunks rather than arbitrary line breaks.

```python
import sys
from pathlib import Path

sys.path.insert(0, str(Path.cwd().parent))

from src.ingestion.chunker import CodebaseChunker

chunker = CodebaseChunker()
chunks = chunker.chunk_file("src/ingestion/loader.py")
print(f"Extracted {len(chunks)} structural AST chunks.")
for c in chunks:
    print(f"[{c.chunk_type.upper()}] {c.qualified_name} (lines {c.start_line}-{c.end_line})")
```
